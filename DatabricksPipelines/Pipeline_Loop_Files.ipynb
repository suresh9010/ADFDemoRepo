```python
dbutils.widgets.text("SourceFilePath","")
dbutils.widgets.text("SinkFilePath","")
dbutils.widgets.text("FileName","")
dbutils.widgets.text("FileFormat","")
```

```python
SourceFilePath = dbutils.widgets.get("SourceFilePath")
SinkFilePath = dbutils.widgets.get("SinkFilePath")
FileName = dbutils.widgets.get("FileName")
FileFormat = dbutils.widgets.get("FileFormat")

def read_write_data(SourceFilePath,SinkFilePath,FileName,FileFormat):
    df = spark.read.format(FileFormat)\
            .option("header",True)\
            .load(SourceFilePath+FileName+'.'+FileFormat)

    df.write.format(FileFormat)\
            .option("header",True)\
            .save(SinkFilePath+FileName)
```

```python
read_write_data(SourceFilePath,SinkFilePath,FileName,FileFormat)
```
